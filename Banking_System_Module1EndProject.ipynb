```python
account_num = '987654'
pin = '4567'
balance = 10000.00


print("        WELCOME TO ABC BANK     ")
print("=" * 40)
```

```python
enter_account = input("Enter Account Number : ")
enter_pin = input("Enter PIN : ")

if enter_account == account_num and enter_pin == pin:

    print('...Login Successful...')

else:
    print("\n ...Login Failed... ")
    print('Invalid Account Number or PIN ')
    print('Please Try Again')
    
```

```python
while True:
     print("\n =========== BANKING MENU =============")

     print("1. Check Balance ")
     print("2. Deposit Money ")
     print("3. Withdraw Money ")
     print("4. Exit ")  


     option = input("\n Enter Your Option (1- 4)::")

     if option == "1":
         print(f"\n Your Current Balance: {balance:.2f} INR")

     elif option == '2':
         amount = input("Enter Amount to be Deposited : ")

         if amount.isdigit():
            
             amount = float(amount)

             if amount > 0:
            
                 balance += amount

                 print("\n Deposit is Successful ")
                 print(f"Updated New Balance: {balance:.2f}")
        
             else:
               print("\n Error: Amount Must be greater than Zero. ")

         else:
            print("\n Error : Amount Must be in Numeric.")

     elif option == '3':
         amount = input("\n Enter the Amount to be Withdraw: ")

         if amount.isdigit():

             amount = float(amount)

             if amount <= 0:
                 print("Error : Amount Must be greater than Zero")

             elif amount > balance:
                 print("Insufficient Balance...")

             else:
                 balance -= amount
                 print("Withdrawal is Successful")
                 print(f"Remaining Blance : {balance:.2f} ")

         else:
             print(" Please Enter a Valid Number ")

     elif option == "4":
         print("Thank You for using ABC Bank!!!")
         break

     else:
         print("Invalid Option. \n Please Select 1 to 4. ")

                    
```
